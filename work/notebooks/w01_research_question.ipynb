### 2) The Question: Decision, Action, Cost of a Wrong Call
* **Unit of Analysis:** A single client page (URL) observed over a 90-day performance window.
* **Search Question:** Which high-value pages have entered structural organic decay and require editorial updates within the next 30 days?
* **Model Output:** A probability score / rank indicating the risk of sustained decline (`is_declining_label`).
* **Decision & Action:** The content team reviews and updates the top 50 flagged URLs each month (e.g., updating outdated facts, optimizing metadata, strengthening internal links).
* **Cost of a Wrong Call:**
  * *False Positive (Wasted Effort):* The team rewrites an evergreen page that would have recovered on its own, wasting content resources.
  * *False Negative (Missed Decay):* A high-traffic revenue page drops from rank 3 to rank 15 unnoticed, causing lost leads and revenue.
* **Why ML over simple rules:** Simple drop rules fail to separate seasonal variations or small query blips from true content decay. ML combines query positions, impressions, CTR, content age, and update history to detect genuine decay patterns.

```python
# Decision framework and operational cost matrix
cost_matrix = {
    "False Positive (Type I)": "Wasted human editorial hours rewriting an evergreen page (~4-6 content hours per URL)",
    "False Negative (Type II)": "Undetected rank degradation from Page 1 to Page 3, causing sustained organic revenue loss",
    "Operational Capacity": "Top 50 high-risk URLs queued for monthly content audit"
}

for scenario, cost in cost_matrix.items():
    print(f"[{scenario}]: {cost}")
```

### 3) Quick Look at the Data (2–3 Real Numbers)
We inspect the dataset generated by the pipeline to quantify the baseline decay rate, identify severe visibility-vs-click gaps, and check age disparity between healthy and decaying pages.

```python
import pandas as pd

# Load the generated dataset directly from the pipeline output
df = pd.read_csv("outputs/refresh_queue.csv")
print(f"Successfully loaded {len(df):,} scored rows from outputs/refresh_queue.csv")

# Compute the 3 required numbers
total_pages = len(df)
declining_count = int(df["is_declining_label"].sum()) if "is_declining_label" in df.columns else int((df["decay_probability"] > 0.5).sum())
decline_rate = declining_count / total_pages

# Engagement / CTR opportunity check
zero_clicks_high_imp = int(len(df[(df["impressions_90d"] > 1000) & (df["clicks_90d"] == 0)]))

# Content age comparison
if "is_declining_label" in df.columns:
    avg_age_decaying = df[df["is_declining_label"] == 1]["content_age_days"].mean()
    avg_age_healthy = df[df["is_declining_label"] == 0]["content_age_days"].mean()
else:
    avg_age_decaying = df[df["decay_probability"] > 0.5]["content_age_days"].mean()
    avg_age_healthy = df[df["decay_probability"] <= 0.5]["content_age_days"].mean()

print("=" * 60)
print(f"Number 1: Starter dataset size is {total_pages:,} pages, with an overall decay rate of {decline_rate:.1%} ({declining_count:,} decaying pages).")
print(f"Number 2: {zero_clicks_high_imp:,} pages have >1,000 impressions but zero clicks, highlighting severe CTR decay opportunities.")
print(f"Number 3: Decaying pages average {avg_age_decaying:.1f} days of content age vs. {avg_age_healthy:.1f} days for healthy pages.")
print("=" * 60)
```
